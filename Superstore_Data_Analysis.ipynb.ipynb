```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px

%matplotlib inline
```

```python
df = pd.read_csv("Sample_Superstore_Cleaned_UTF8.csv")
```

```python
df["Postal Code"] = df["Postal Code"].fillna(df["Postal Code"].mode()[0])
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()

print("Duplicate Records:", df.duplicated().sum())
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"], errors="coerce")
df["Ship Date"] = pd.to_datetime(df["Ship Date"], errors="coerce")

df[["Order Date", "Ship Date"]].head()
```

```python
df["Order Year"] = df["Order Date"].dt.year
df["Order Month"] = df["Order Date"].dt.month_name()
df["Order Day"] = df["Order Date"].dt.day_name()

df.head()
```

```python
df.info()
```

```python
total_sales = df["Sales"].sum()

print("Total Sales :", round(total_sales,2))
```

```python
total_profit = df["Profit"].sum()

print("Total Profit :", round(total_profit,2))
```

```python
average_sales = df["Sales"].mean()

print("Average Sales :", round(average_sales,2))
```

```python
average_profit = df["Profit"].mean()

print("Average Profit :", round(average_profit,2))
```

```python
print("Total Orders :", df["Order ID"].nunique())
print("Total Customers :", df["Customer ID"].nunique())
print("Total Products :", df["Product Name"].nunique())
```

```python
sales_category = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

sales_category
```

```python
sales_category.plot(kind="bar", figsize=(8,5))

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")

plt.xticks(rotation=0)

plt.show()
```

```python
profit_category = df.groupby("Category")["Profit"].sum().sort_values(ascending=False)

profit_category
```

```python
profit_category.plot(kind="bar", figsize=(8,5))

plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit")

plt.xticks(rotation=0)

plt.show()
```

```python
segment_sales = df.groupby("Segment")["Sales"].sum().sort_values(ascending=False)

segment_sales
```

```python
segment_sales.plot(kind="pie", autopct="%1.1f%%", figsize=(7,7))

plt.title("Sales by Segment")

plt.ylabel("")

plt.show()
```

```python
df.to_csv("Retail_Cleaned_Dataset.csv", index=False)

print("Cleaned Dataset Exported Successfully")
```
